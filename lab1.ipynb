```python
import cv2
import glob as gl
import random
```

```python
base_path = "/content/drive/MyDrive/fac/Laborator/Images"
images = []

for i in range(4):
  images.append(gl.glob(base_path+'/'+str(i)+'/*.jpg'))
print(images[0])
```

```python
random.seed(42)

random_list = random.sample(range(0,9),7)

train_images = []
test_images = []

for idx in range(4):
  tmp_train = []
  tmp_test = []
  for i in range(10):
    if i in random_list:
      tmp_train.append(images[idx][i])
    else:
      tmp_test.append(images[idx][i])
  train_images.append(tmp_train)
  test_images.append(tmp_test)

print(random_list)
print(test_images)
```

```python
tr_im = []
ts_im = []

for i in range(len(train_images)):
  tmp = []
  for j in range(len(train_images[0])):
    im = cv2.imread(train_images[i][j])
    tmp.append(im)
  tr_im.append(tmp)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(tr_im[0][0])
```

```python
# RGB BGR YUV etc
```

```python
for i in range(len(tr_im)):
  for j in range(len(tr_im[0])):
    tr_im[i][j] = cv2.cvtColor(tr_im[i][j], cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(tr_im[0][0])
```

```python
scale_precent = 0.15
width = int(tr_im[0][0].shape[1] * scale_precent)
height = int(tr_im[0][0].shape[0] * scale_precent)
dim = (width, height)

for i in range(len(tr_im)):
  for j in range(len(tr_im[0])):
    tr_im[i][j] = cv2.resize(tr_im[i][j],dim)
```
